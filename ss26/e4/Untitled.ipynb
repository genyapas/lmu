```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import assoc_laguerre, factorial
```

```python
def radial_fkt(n, l, r_a0):
    prefix = -np.sqrt((2/n)**3 * factorial(n-l-1) / (2*n * (factorial(n+l))**3))    
    rho = 2 * r_a0 / n
    laguerre = assoc_laguerre(rho, n-l-1, 2*l+1)
    return prefix * laguerre * rho**l * np.exp(-rho/2)
```

```python
r = np.linspace(0, 20, 500)
```

```python
fig, ax = plt.subplots()
ax.plot(r, radial_fkt(1, 0, r,), label='1s')
ax.legend()
plt.savefig('n1l0.png')
```
